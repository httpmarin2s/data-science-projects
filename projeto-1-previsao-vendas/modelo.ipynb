# **[1] Análise Inicial**

```python
def analise_inicial(df):
    print("-"*60)
    print("Informações iniciais")
    print(df.info())

    print("-"*60)
    print("Total de Linhas x Colunas")
    print(df.shape)

    print("-"*60)
    print("Total de valores nulos")
    print(df.isna().sum())

    print("-"*60)
    print("Valors únicos")
    print(df.nunique())

    print("-"*60)
    print("Análise de categóricas")
    for col in df.select_dtypes(include='object').columns:
        print(df[col].value_counts())
    
    print("-"*60)
    print("Estatísticas Iniciais")
    print(df.describe(include='all'))
```

```python
analise_inicial(df)
```
