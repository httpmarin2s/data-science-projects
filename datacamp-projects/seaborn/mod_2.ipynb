# **Preparando o ambiente**

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# simulando dados de poluição do ar
np.random.seed(42) #pegar os mesmos dados sempre

horas = np.arange(0,24) #cria valores inteiros de 0 a 23

poluicao = np.random.normal(loc=50, scale=10, size=24)
# gera valores de uma distribuição normal
# média 10
# sesvio padrão = 10
# 24 itens

poluicao = np.clip(poluicao, 20,100)
# geração de um intervalo minimo e máximo
# 20-100
```

```python
df_poluicao = pd.DataFrame({
    'Hora': horas,
    'PM2': poluicao
})
```

```python
np.random.seed(42)
potencia = np.random.randint(70, 300, size=50)
consumo = np.random.normal(15, 3, size=50) - (potencia/100)
consumo = np.clip(consumo, 5, 20)
cilindros = np.random.randint(4,8,size=50)
```

```python
df_carros = pd.DataFrame({
    'Potencia':potencia, 
    'Consumo': consumo, 
    'Cilindros': cilindros,
    'Fabricante': np.random.choice(['Ford', 'Chevrolet', 'Toyota', 'BMW'], size=50)
})
```

```python
df_poluicao.head()
```

```python
df_carros.head()
```

## **[1] Gráfico de dispersão simples**
- Usando df_carros, crie um gráfico de dispersão relacionando Potencia_HP (eixo X) e Consumo_L_100km (eixo Y) com Seaborn. 

```python
sns.relplot(
    kind='scatter',
    data=df_carros,
    x='Potencia', 
    y='Consumo'
)

plt.show()
```

## **[2] Personalizando cores no scatterplot**
- Refaça o gráfico anterior, mas agora colorindo os pontos pelo fabricante (Fabricante).

```python
sns.relplot(
    kind='scatter',
    data=df_carros,
    x='Potencia', 
    y='Consumo',
    hue='Fabricante'
)

plt.show()
```

## **[3] Alterando o tamanho dos pontos**
- No mesmo gráfico, faça com que o tamanho dos pontos seja proporcional à potência do carro.

```python
sns.relplot(
    kind='scatter',
    data=df_carros,
    x='Potencia', 
    y='Consumo',
    hue='Fabricante',
    size='Potencia'
)

plt.show()
```

## **[4] Alterando o estilo dos pontos**
- Personalize o scatterplot para que cada fabricante tenha formato de marcador diferente (style).

```python
sns.relplot(
    kind='scatter',
    data=df_carros,
    x='Potencia', 
    y='Consumo',
    hue='Fabricante',
    size='Potencia',
    style='Fabricante'
)

plt.show()
```

## **[5] Criando subgráficos (col e row)**
- Crie subgráficos com col='Fabricante' para mostrar como a relação entre potência e consumo varia por fabricante.

```python
sns.relplot(
    kind='scatter',
    data=df_carros,
    x='Potencia', 
    y='Consumo',
    hue='Fabricante',
    size='Potencia',
    col='Fabricante'
)

plt.show()
```

## **[6] Criando subgráficos com dois fatores**
Usando row='Fabricante' e col='Cilindros' (crie uma nova coluna com valores aleatórios 4, 6 ou 8 cilindros), gere múltiplos scatterplots.

```python
sns.relplot(
    kind='scatter',
    data=df_carros,
    x='Potencia', 
    y='Consumo',
    hue='Fabricante',
    col='Cilindros'
)

plt.show()
```

## **[7] Introdução a gráfico de linhas**
- Usando df_poluicao, crie um lineplot mostrando como o nível de poluição (PM2.5) varia ao longo do dia (Hora).

```python
df_poluicao.head()
```

```python
sns.relplot(
    kind='line',
    data=df_poluicao,
    x='Hora', 
    y='PM2'
)

plt.show()
```

## **[8] Interpretando gráfico de linhas**
- Inclua marcadores nos pontos do gráfico de linhas e adicione rótulos aos eixos para facilitar a leitura.

```python
sns.relplot(
    kind='line',
    data=df_poluicao,
    x='Hora', 
    y='PM2',
    markers=True
)

plt.show()
```
