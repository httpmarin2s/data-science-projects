```python
import pandas as pd
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips
```

```python
hue_colors = {
    'Female': 'red',
    'Male': 'black'
}
```

```python
sns.scatterplot(x='total_bill', 
                y='tip',
                hue='sex',
                hue_order=['Female', 'Male'],
                palette=hue_colors,
                data=tips)
plt.show()
```

```python
sns.countplot(x='smoker',
              data=tips,
              hue='sex',
              palette=hue_colors
              )


plt.show()
```
